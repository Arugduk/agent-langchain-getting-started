# 1.系统提示词
System Message

```python
from langchain.agents import create_agent
from langchain.messages import HumanMessage
from dotenv import load_dotenv
from langchain.chat_models import init_chat_model
import os

load_dotenv()
base_url = "https://apinebula.ai/v1"
api_key = os.getenv("OPENAI_API_KEY")

model = init_chat_model(model="gpt-5.6-sol",
                        base_url= base_url,
                        api_key=api_key)
agent = create_agent(
    model =model,
    system_prompt="你以祖国人的口吻来回答问题"

)

for token,metadata in agent.stream(
        {
            "messages": [HumanMessage(content="你是谁?")]
        },
    stream_mode="messages"
):
    print(token.content,end="",flush=True)
```

# 2.1设定角色和指令

```python
system_prompt ="""
你是一个编程助手,你帮助用户编写代码
"""

agent = create_agent(
    model =model,
    system_prompt=system_prompt
)
for token,metadata in agent.stream(
    {
            "messages": [HumanMessage(content="怎么定义string变量来记录学校名字")],
    },
    stream_mode="messages"
):
    content = token.content
    # 新版：content 是块列表
    if isinstance(content, list):
        for block in content:
            if isinstance(block, dict) and block.get("type") == "text":
                text = block.get("text", "")
                # 只保留最终答案，跳过推理
                if block.get("phase") in (None, "final_answer") and text:
                    print(text, end="", flush=True)
    # 旧版：content 是字符串
    elif isinstance(content, str):
        print(content, end="", flush=True)
```

```python
system_prompt="""
# 身份
- 你是一个编程助手,你帮助用户编写代码
# 指令
- 定义变量时,使用snake case 命名法,不要用camel case
- 不要返回markdown格式说明,仅仅返回代码即可
"""
agent = create_agent(
    model =model,
    system_prompt=system_prompt
)

for token,metadata in agent.stream(
    {
            "messages": [HumanMessage(content="怎么定义string变量来记录学校名字")],
    },
    stream_mode="messages"
):
    content = token.content
    # 新版：content 是块列表
    if isinstance(content, list):
        for block in content:
            if isinstance(block, dict) and block.get("type") == "text":
                text = block.get("text", "")
                # 只保留最终答案，跳过推理
                if block.get("phase") in (None, "final_answer") and text:
                    print(text, end="", flush=True)
    # 旧版：content 是字符串
    elif isinstance(content, str):
        print(content, end="", flush=True)
```

# 2.3 结构化输出
## a.基于提示词的结构化输出

```python
system_prompt="""
#身份
一你是一个科幻作家，根据用户的要求创建一个太空之都
#指令
请务必以JS0N格式输出，不要加任何markdown样式。
#示例：
user:月球的首都是什么？
assistant:
{
    "name":"月华市(Lunaria)",
    "location":"位于月球正面赤道附近的静海基地遗址之上，依托巨大的穹顶与地下网络建成”,
    "vibe":"冷洌、高效、革新",
    "economy":"氨-3能源开采、量子通信枢纽、尖端生物圈农业"
}
"""
agent = create_agent(
    model =model,
    system_prompt=system_prompt
)

for token,metadata in agent.stream(
    {
            "messages": [HumanMessage(content="金星的首都是什么")],
    },
    stream_mode="messages"
):
    content = token.content
    # 新版：content 是块列表
    if isinstance(content, list):
        for block in content:
            if isinstance(block, dict) and block.get("type") == "text":
                text = block.get("text", "")
                # 只保留最终答案，跳过推理
                if block.get("phase") in (None, "final_answer") and text:
                    print(text, end="", flush=True)
    # 旧版：content 是字符串
    elif isinstance(content, str):
        print(content, end="", flush=True)
```

## b.基于Model的结构化输出

```python
from pydantic import BaseModel

class CapitalInfo(BaseModel):
    name: str
    location: str
    vibe: str
    economy: str
```

```python
system_prompt="""
#身份
一你是一个科幻作家，根据用户的要求创建一个太空之都
"""

agent = create_agent(
    model =model,
    system_prompt=system_prompt,
    response_format= CapitalInfo
)
response = agent.invoke(
    {
            "messages": [HumanMessage(content="金星的首都是什么")],
    }
)
print(response)

```
