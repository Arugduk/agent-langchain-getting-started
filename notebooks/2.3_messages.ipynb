```python
from langchain.agents import create_agent
from langchain.tools import tool
from langchain.chat_models import init_chat_model
import os
from dotenv import load_dotenv
from langchain_core.messages import SystemMessage, HumanMessage

load_dotenv()
#定义工具
@tool
def get_weather(location:str)->str:
    """

    :param location:
    :return:
    """
    return f"current weather in {location} is sunny"

# 调用init_chat_model函数初始化模型
# 参数model用来指定模型名称，Langchain会根据模型名字自动设定base_url，并从环境变量中获取api_key
base_url = "https://apinebula.ai/v1"
api_key = os.getenv("OPENAI_API_KEY")

model = init_chat_model(model="gpt-5.6-sol",
                        base_url= base_url,
                        api_key=api_key)
#创建agent
agent=create_agent(model,tools=[get_weather])

response=agent.invoke({
    "messages": [
        #{"role":"system","content":"你是一个热心的ai助手"},
        #{"role":"user","content":"今天天气怎么样？"}
        SystemMessage(content="你是一个热心的ai助手"),
        HumanMessage(content="今天苏州天气怎么样？")
    ]
})

print(response)

```

```python
for message in response['messages']:
    message.pretty_print()
```

```python
message = HumanMessage([
    {"type":"text","text":"你好"},
])
```

```python
stream=agent.stream(
    {"messages":[message]},
    stream_mode="messages"
)

for chunk, metadata in stream:
    print(chunk.content, end="", flush=True)
```
