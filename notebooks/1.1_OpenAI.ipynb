# 1.初始化客户端
## 1.1 硬编码API_KEY

```python
import os
from openai import OpenAI
from dotenv import load_dotenv

load_dotenv()

# 创建openAI客户端
client = OpenAI(
    api_key=os.environ["OPENAI_API_KEY"],
    base_url="https://apinebula.ai/v1",
)
```

## 1.2 从环境变量获取api_key
用到python-dotenv

```python
from dotenv import load_dotenv
import os

load_dotenv()

api_key=os.getenv("OPENAI_API_KEY")

client = OpenAI(
    api_key= api_key,
    base_url="https://apinebula.ai/v1",
)
```

# 2.访问模型

```python
print("正在调用")
response = client.chat.completions.create(
    messages = [
        {"role":"system","content":"你是一名友好的助教"},
        {"role":"user","content":"你是谁"}
    ],
    model ="gpt-5.6-sol",
    stream=False
)
print(response)
```

# 3.获取结果

```python
print(response.model_dump_json())
```
