# 1.加载环境变量

```python
from dotenv import load_dotenv
import os
load_dotenv()
api_key=os.getenv("OPENAI_API_KEY")
#这里不用os.getenv,langchain会自动做这件事
```

# 2.定义工具

```python
from langchain.tools import tool
#需要描述
@tool
def getWeather(location: str)->str:
    """

    :param location:
    :return:
    """

    return f"Current weather in {location} is sunny"
```

# 3.创建Agent

```python
from langchain.agents import create_agent
from langchain_openai import  ChatOpenAI
# 1. 先创建模型对象，在这里配置 base_url 和 api_key
model = ChatOpenAI(
    model="gpt-5.6-sol",
    api_key= api_key,
    base_url="https://apinebula.ai/v1",  # 关键：在这里配置
)
agent = create_agent(
    model=model,

    tools = [getWeather],

)
```

# 4.发起调用

```python
print("正在调用")
response =agent.invoke({
    "messages": [
        {"role": "user","content":"苏州今天天气如何？"}
    ]
})
print(response)
```
